```python
import hashlib, json, time

class Block:
    def __init__(self, index, data, previous_hash):
        self.index = index
        self.timestamp = time.time()
        self.data = data
        self.previous_hash = previous_hash
        self.hash = self.calculate_hash()

    def calculate_hash(self):
        payload = {
            "index": self.index,
            "timestamp": self.timestamp,
            "data": self.data,
            "previous_hash": self.previous_hash,
        }
        raw = json.dumps(payload, sort_keys=True).encode()
        return hashlib.sha256(raw).hexdigest()

class Blockchain:
    def __init__(self):
        self.chain = [self.create_genesis_block()]

    def create_genesis_block(self):
        # TODO: buat genesis block
        return Block(0, "Genesis Block", "0")

    def latest_block(self):
        return self.chain[-1]

    def add_block(self, data):
        # TODO: buat block baru dengan previous_hash dari latest block
        new_block = Block(self.latest_block().index + 1, data, self.latest_block().hash)
        self.chain.append(new_block)

    def is_valid(self):
        # TODO: cek hash block dan previous_hash link untuk seluruh chain
        if self.chain[0].hash != self.chain[0].calculate_hash():
            return False
        for i in range(1, len(self.chain)):
            if self.chain[i].hash != self.chain[i].calculate_hash():
                return False
            if self.chain[i].previous_hash != self.chain[i - 1].hash:
                return False
        return True
```

```python
# Skenario 1: Chain baru hanya memiliki genesis block (Expected: Valid)
bc = Blockchain()
print("Jumlah block:", len(bc.chain))
print("Apakah valid?:", bc.is_valid())
```

```python
# Skenario 2: Tambah minimal 3 block dengan transaksi berbeda (Expected: Valid)
bc.add_block("Transaksi 1: Alice kirim 10 koin ke Bob")
bc.add_block("Transaksi 2: Bob kirim 5 koin ke Charlie")
bc.add_block("Transaksi 3: Charlie kirim 2 koin ke David")

print("Jumlah block:", len(bc.chain))
print("Apakah valid?:", bc.is_valid())
```

```python
# Skenario 3: Ubah data pada block tengah tanpa menghitung ulang chain (Expected: Tidak valid)
data_asli = bc.chain[2].data
bc.chain[2].data = "Transaksi Palsu: Bob kirim 1000 koin ke Charlie"

print("Data block index 2 diubah menjadi:", bc.chain[2].data)
print("Apakah valid?:", bc.is_valid())

# Kembalikan ke data asli agar pengujian skenario 4 & 5 tidak terpengaruh
bc.chain[2].data = data_asli
```

```python
# Skenario 4: Ubah previous_hash pada block (Expected: Tidak valid)
prev_hash_asli = bc.chain[2].previous_hash
bc.chain[2].previous_hash = "00000000000000000000000000000000"

print("previous_hash block index 2 diubah menjadi:", bc.chain[2].previous_hash)
print("Apakah valid?:", bc.is_valid())

# Kembalikan ke previous_hash asli agar cetakan pada skenario 5 tetap menampilkan keterkaitan asli
bc.chain[2].previous_hash = prev_hash_asli
```

```python
# Skenario 5: Cetak index, hash, dan previous_hash setiap block (Expected: Keterkaitan terlihat)
for block in bc.chain:
    print(f"Index         : {block.index}")
    print(f"Data          : {block.data}")
    print(f"Previous Hash : {block.previous_hash}")
    print(f"Hash          : {block.hash}")
    print("-" * 60)
```
